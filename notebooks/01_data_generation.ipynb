```python
import pandas as pd
import numpy as np

print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
```

```python
np.random.seed(42)

print("Environment setup successful!")
```

```python
n = 1000

student_ids = [f"STU{str(i).zfill(4)}" for i in range(1, n + 1)]

print("Number of students:", len(student_ids))
print("First 5 IDs:", student_ids[:5])
```

```python
gender = np.random.choice(
    ["Male", "Female"],
    size=n,
    p=[0.55, 0.45]
)

age = np.random.randint(20, 24, size=n)

department = np.random.choice(
    ["CSE", "ECE", "EEE", "MECH", "CIVIL", "IT", "DS"],
    size=n
)

graduation_year = np.random.choice(
    [2026, 2027],
    size=n,
    p=[0.35, 0.65]
)

print("Gender:", gender[:5])
print("Age:", age[:5])
print("Department:", department[:5])
print("Graduation Year:", graduation_year[:5])
```

```python
tenth_percentage = np.round(np.random.normal(78, 10, n).clip(50, 100), 2)

twelfth_percentage = np.round(np.random.normal(75, 11, n).clip(45, 100), 2)

cgpa = np.round(np.random.normal(7.6, 0.9, n).clip(5.0, 10.0), 2)

attendance_percentage = np.round(
    np.random.normal(82, 9, n).clip(55, 100),
    2
)

backlogs = np.random.choice(
    [0, 1, 2, 3, 4],
    size=n,
    p=[0.65, 0.18, 0.09, 0.05, 0.03]
)

print("10th Percentage:", tenth_percentage[:5])
print("12th Percentage:", twelfth_percentage[:5])
print("CGPA:", cgpa[:5])
print("Attendance:", attendance_percentage[:5])
print("Backlogs:", backlogs[:5])
```

```python
coding_skill = np.round(
    np.random.normal(65, 15, n).clip(20, 100),
    2
)

communication_skill = np.round(
    np.random.normal(68, 14, n).clip(20, 100),
    2
)

technical_skill = np.round(
    np.random.normal(70, 13, n).clip(20, 100),
    2
)

print("Coding Skill:", coding_skill[:5])
print("Communication Skill:", communication_skill[:5])
print("Technical Skill:", technical_skill[:5])
```

```python
aptitude_score = np.round(
    np.random.normal(65, 15, n).clip(20, 100),
    2
)

print("Aptitude Score:", aptitude_score[:5])
```

```python
internships = np.random.choice(
    [0, 1, 2, 3],
    size=n,
    p=[0.35, 0.40, 0.20, 0.05]
)

projects = np.random.choice(
    [1, 2, 3, 4, 5],
    size=n,
    p=[0.10, 0.25, 0.30, 0.25, 0.10]
)

print("Internships:", internships[:5])
print("Projects:", projects[:5])
```

```python
certifications = np.random.choice(
    [0, 1, 2, 3, 4, 5],
    size=n,
    p=[0.10, 0.20, 0.25, 0.20, 0.15, 0.10]
)

hackathons = np.random.choice(
    [0, 1, 2, 3],
    size=n,
    p=[0.55, 0.30, 0.10, 0.05]
)

print("Certifications:", certifications[:5])
print("Hackathons:", hackathons[:5])
```

```python
placement_preparation_hours = np.round(
    np.random.normal(8, 4, n).clip(0, 25),
    1
)

mock_interviews = np.random.choice(
    [0, 1, 2, 3, 4, 5, 6],
    size=n,
    p=[0.10, 0.15, 0.20, 0.20, 0.15, 0.10, 0.10]
)

resume_score = np.round(
    np.random.normal(68, 14, n).clip(30, 100),
    2
)

print("Preparation Hours:", placement_preparation_hours[:5])
print("Mock Interviews:", mock_interviews[:5])
print("Resume Score:", resume_score[:5])
```

```python
interview_score = np.round(
    np.random.normal(67, 15, n).clip(20, 100),
    2
)

print("Interview Score:", interview_score[:5])
```

```python
placement_score = (
    0.20 * (cgpa / 10 * 100)
    + 0.10 * attendance_percentage
    + 0.10 * coding_skill
    + 0.10 * communication_skill
    + 0.10 * technical_skill
    + 0.10 * aptitude_score
    + 5 * internships
    + 3 * projects
    + 1.5 * certifications
    + 2 * hackathons
    + 1.5 * placement_preparation_hours
    + 2 * mock_interviews
    + 0.05 * resume_score
    + 0.10 * interview_score
    - 4 * backlogs
    + np.random.normal(0, 10, n)
)

placement_status = np.where(
    placement_score >= np.percentile(placement_score, 60),
    "Placed",
    "Not Placed"
)

print("Placed:", np.sum(placement_status == "Placed"))
print("Not Placed:", np.sum(placement_status == "Not Placed"))
```

```python
package_lpa = np.where(
    placement_status == "Placed",
    np.round(
        3
        + 0.35 * cgpa
        + 0.015 * coding_skill
        + 0.015 * technical_skill
        + 0.02 * aptitude_score
        + 0.10 * internships
        + 0.05 * projects
        + 0.03 * interview_score
        + np.random.normal(0, 1.2, n),
        2
    ),
    0
)

package_lpa = np.round(
    np.clip(package_lpa, 0, 20),
    2
)

print("Package LPA:", package_lpa[:10])
```

```python
df = pd.DataFrame({
    "Student_ID": student_ids,
    "Gender": gender,
    "Age": age,
    "Department": department,
    "Graduation_Year": graduation_year,
    "10th_Percentage": tenth_percentage,
    "12th_Percentage": twelfth_percentage,
    "CGPA": cgpa,
    "Attendance_Percentage": attendance_percentage,
    "Backlogs": backlogs,
    "Coding_Skill": coding_skill,
    "Communication_Skill": communication_skill,
    "Technical_Skill": technical_skill,
    "Aptitude_Score": aptitude_score,
    "Internships": internships,
    "Projects": projects,
    "Certifications": certifications,
    "Hackathons": hackathons,
    "Placement_Preparation_Hours": placement_preparation_hours,
    "Mock_Interviews": mock_interviews,
    "Resume_Score": resume_score,
    "Interview_Score": interview_score,
    "Placement_Status": placement_status,
    "Package_LPA": package_lpa
})

print("Dataset shape:", df.shape)
```

```python
with engine.connect() as connection:
    result = connection.exec_driver_sql(
        "SELECT COUNT(*) FROM student_performance"
    )
    print("Rows in SQL table:", result.fetchone()[0])
```

```python
query = """
SELECT
    COUNT(*) AS total_students,
    SUM(CASE WHEN Placement_Status = 'Placed' THEN 1 ELSE 0 END) AS placed_students,
    ROUND(
        100.0 * SUM(CASE WHEN Placement_Status = 'Placed' THEN 1 ELSE 0 END) / COUNT(*),
        2
    ) AS placement_rate
FROM student_performance;
"""

result = pd.read_sql_query(query, engine)

result
```

```python
query = """
SELECT
    Department,
    COUNT(*) AS total_students,
    SUM(CASE WHEN Placement_Status = 'Placed' THEN 1 ELSE 0 END) AS placed_students,
    ROUND(
        100.0 * SUM(CASE WHEN Placement_Status = 'Placed' THEN 1 ELSE 0 END) / COUNT(*),
        2
    ) AS placement_rate
FROM student_performance
GROUP BY Department
ORDER BY placement_rate DESC;
"""

department_result = pd.read_sql_query(query, engine)

department_result
```

```python
query = """
SELECT
    Placement_Status,
    ROUND(AVG(CGPA), 2) AS average_cgpa,
    ROUND(AVG(Attendance_Percentage), 2) AS average_attendance,
    ROUND(AVG(Coding_Skill), 2) AS average_coding_skill,
    ROUND(AVG(Communication_Skill), 2) AS average_communication_skill,
    ROUND(AVG(Technical_Skill), 2) AS average_technical_skill
FROM student_performance
GROUP BY Placement_Status;
"""

skill_result = pd.read_sql_query(query, engine)

skill_result
```

```python
query = """
SELECT
    Internships,
    COUNT(*) AS total_students,
    SUM(CASE WHEN Placement_Status = 'Placed' THEN 1 ELSE 0 END) AS placed_students,
    ROUND(
        100.0 * SUM(CASE WHEN Placement_Status = 'Placed' THEN 1 ELSE 0 END) / COUNT(*),
        2
    ) AS placement_rate
FROM student_performance
GROUP BY Internships
ORDER BY Internships;
"""

internship_result = pd.read_sql_query(query, engine)

internship_result
```

```python
query = """
SELECT
    Projects,
    COUNT(*) AS total_students,
    SUM(CASE WHEN Placement_Status = 'Placed' THEN 1 ELSE 0 END) AS placed_students,
    ROUND(
        100.0 * SUM(CASE WHEN Placement_Status = 'Placed' THEN 1 ELSE 0 END) / COUNT(*),
        2
    ) AS placement_rate
FROM student_performance
GROUP BY Projects
ORDER BY Projects;
"""

project_result = pd.read_sql_query(query, engine)

project_result
```

```python
query = """
SELECT
    Backlogs,
    COUNT(*) AS total_students,
    SUM(CASE WHEN Placement_Status = 'Placed' THEN 1 ELSE 0 END) AS placed_students,
    ROUND(
        100.0 * SUM(CASE WHEN Placement_Status = 'Placed' THEN 1 ELSE 0 END) / COUNT(*),
        2
    ) AS placement_rate
FROM student_performance
GROUP BY Backlogs
ORDER BY Backlogs;
"""

backlog_result = pd.read_sql_query(query, engine)

backlog_result
```

```python
query = """
SELECT
    Certifications,
    COUNT(*) AS total_students,
    SUM(CASE WHEN Placement_Status = 'Placed' THEN 1 ELSE 0 END) AS placed_students,
    ROUND(
        100.0 * SUM(CASE WHEN Placement_Status = 'Placed' THEN 1 ELSE 0 END) / COUNT(*),
        2
    ) AS placement_rate
FROM student_performance
GROUP BY Certifications
ORDER BY Certifications;
"""

certification_result = pd.read_sql_query(query, engine)

certification_result
```

```python
query = """
SELECT
    Hackathons,
    COUNT(*) AS total_students,
    SUM(CASE WHEN Placement_Status = 'Placed' THEN 1 ELSE 0 END) AS placed_students,
    ROUND(
        100.0 * SUM(CASE WHEN Placement_Status = 'Placed' THEN 1 ELSE 0 END) / COUNT(*),
        2
    ) AS placement_rate
FROM student_performance
GROUP BY Hackathons
ORDER BY Hackathons;
"""

hackathon_result = pd.read_sql_query(query, engine)

hackathon_result
```

```python
query = """
SELECT
    CASE
        WHEN Placement_Preparation_Hours <= 5 THEN '0-5'
        WHEN Placement_Preparation_Hours <= 10 THEN '6-10'
        WHEN Placement_Preparation_Hours <= 15 THEN '11-15'
        WHEN Placement_Preparation_Hours <= 20 THEN '16-20'
        ELSE '21-25'
    END AS preparation_group,
    COUNT(*) AS total_students,
    SUM(CASE WHEN Placement_Status = 'Placed' THEN 1 ELSE 0 END) AS placed_students,
    ROUND(
        100.0 * SUM(CASE WHEN Placement_Status = 'Placed' THEN 1 ELSE 0 END) / COUNT(*),
        2
    ) AS placement_rate
FROM student_performance
GROUP BY preparation_group
ORDER BY
    CASE preparation_group
        WHEN '0-5' THEN 1
        WHEN '6-10' THEN 2
        WHEN '11-15' THEN 3
        WHEN '16-20' THEN 4
        WHEN '21-25' THEN 5
    END;
"""

prep_result = pd.read_sql_query(query, engine)

prep_result
```

```python
query = """
SELECT
    Mock_Interviews,
    COUNT(*) AS total_students,
    SUM(CASE WHEN Placement_Status = 'Placed' THEN 1 ELSE 0 END) AS placed_students,
    ROUND(
        100.0 * SUM(CASE WHEN Placement_Status = 'Placed' THEN 1 ELSE 0 END) / COUNT(*),
        2
    ) AS placement_rate
FROM student_performance
GROUP BY Mock_Interviews
ORDER BY Mock_Interviews;
"""

mock_result = pd.read_sql_query(query, engine)

mock_result
```

```python
query = """
SELECT
    Department,
    COUNT(*) AS placed_students,
    ROUND(AVG(Package_LPA), 2) AS average_package_lpa
FROM student_performance
WHERE Placement_Status = 'Placed'
GROUP BY Department
ORDER BY average_package_lpa DESC;
"""

package_department_result = pd.read_sql_query(query, engine)

package_department_result
```

```python
query = """
WITH ranked_packages AS (
    SELECT
        Package_LPA,
        ROW_NUMBER() OVER (ORDER BY Package_LPA) AS row_num,
        COUNT(*) OVER () AS total_rows
    FROM student_performance
    WHERE Placement_Status = 'Placed'
)
SELECT
    ROUND(AVG(Package_LPA), 2) AS median_package_lpa
FROM ranked_packages
WHERE row_num IN (
    (total_rows + 1) / 2,
    (total_rows + 2) / 2
);
"""

median_package_result = pd.read_sql_query(query, engine)

median_package_result
```

```python
query = """
SELECT
    ROUND(
        (
            SUM((CGPA - avg_cgpa) * (Package_LPA - avg_package))
        )
        /
        SQRT(
            SUM((CGPA - avg_cgpa) * (CGPA - avg_cgpa))
            *
            SUM((Package_LPA - avg_package) * (Package_LPA - avg_package))
        ),
        4
    ) AS cgpa_package_correlation
FROM (
    SELECT
        CGPA,
        Package_LPA,
        AVG(CGPA) OVER () AS avg_cgpa,
        AVG(Package_LPA) OVER () AS avg_package
    FROM student_performance
    WHERE Placement_Status = 'Placed'
);
"""

cgpa_package_result = pd.read_sql_query(query, engine)

cgpa_package_result
```
